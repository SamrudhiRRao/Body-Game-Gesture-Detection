```python
import cv2
import time
import os

# ----- SETTINGS -----
OUTPUT_DIR = "C:\\Users\\LOQ\\OneDrive\\Desktop\\codes\\Applied AI Engineering Lab\\Body-Game-Gesture-Detection\\gesture_videos"
GESTURES = ["jump", "duck", "left", "right", "neutral"]
NUM_SAMPLES = 3         # how many videos per gesture
VIDEO_DURATION = 5      # seconds
FPS = 30                # frames per second
# --------------------

def record_video(filename, duration=5, fps=30):
    cap = cv2.VideoCapture(0)
    if not cap.isOpened():
        print("ERROR: Could not access webcam.")
        return

    # Get camera resolution
    width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
    height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

    fourcc = cv2.VideoWriter_fourcc(*"mp4v")
    out = cv2.VideoWriter(filename, fourcc, fps, (width, height))

    print(f"\nRecording: {filename}")
    print("Starting in 3 seconds... get ready!")
    time.sleep(3)

    start = time.time()
    frame_count = 0

    while True:
        ret, frame = cap.read()
        if not ret:
            break

        out.write(frame)
        frame_count += 1

        # Show preview
        cv2.imshow("Recording...", frame)
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

        if time.time() - start >= duration:
            break

    cap.release()
    out.release()
    cv2.destroyAllWindows()
    print(f"Saved: {filename} ({frame_count} frames)")


def main():
    # Create folder if not exists
    os.makedirs(OUTPUT_DIR, exist_ok=True)

    for gesture in GESTURES:
        print(f"\n==== Recording gesture: {gesture} ====")
        for i in range(1, NUM_SAMPLES + 1):
            filename = os.path.join(OUTPUT_DIR, f"{gesture}_{i}.mp4")
            record_video(filename, VIDEO_DURATION, FPS)

    print("\n🎉 All gesture videos recorded!")


if __name__ == "__main__":
    main()
```
